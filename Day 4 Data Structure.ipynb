Q14 Find the second largest salary.

```python
salary = [50000,75000,62000,90000,85000]
```

```python
largest = salary[0]
second_largest = float("-inf")

for x in salary:
    if x > largest:
        second_largest = largest
        largest = x

    elif x > second_largest and x != largest:
        second_largest = x

print(second_largest)
```

Q16 You have transaction IDs:

Write a function that:

* Removes duplicates
* Preserves the original order
* Returns the cleaned list

```python
transactions = [101,102,103,101,105,102,106,107,106]
```

```python
unique = []

for x in transactions:
    if x not in unique:
        unique.append(x)

print(unique)
```

```python
def remove_dup(t_list):
    
    unique = []

    for x in t_list:
        if x not in unique:
            unique.append(x)

    return unique
```

```python
remove_dup(transactions)
```

Q5. Check if a String is a Palindrome

```python
text = 'madam'

reverse = ''

for x in text:
    reverse = x + reverse
if text == reverse:
    print('palindrome')

else:
    print('not palindrome')
```

```python
review = "OpenAI GPT5 is Amazing 2026!!!"
```

```python
upper_count = 0
lower_count = 0
digit_count = 0
space_count = 0
special_count = 0

for x in review:
    if x.isupper():
        upper_count += 1
    elif x.islower():
        lower_count += 1
    elif x.isdigit():
        digit_count += 1
    elif x.isspace():
        space_count += 1
    else:
        special_count += 1
        
print(f"Uppercase: {upper_count}")
print(f"Lowercase: {lower_count}")
print(f"Digits: {digit_count}")
print(f"Spaces: {space_count}")
print(f"Special characters: {special_count}")
```

Q9: Find the character that appears the most.

```python
text = "banana"
```

```python
frequency = {}

for x in text:
    if x in frequency:
        frequency[x] += 1
    else:
        frequency[x] = 1

max_y = None
max_count = 0

for y in frequency:
    if frequency[y] > max_count:
        max_count = frequency[y]
        max_y = y
print(max_y)
```
